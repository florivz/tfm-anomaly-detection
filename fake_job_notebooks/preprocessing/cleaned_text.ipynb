# Fake Jobs – Cleaned + Freitexte
- Cleaned-Features plus die 5 originalen Freitextspalten (für AnoLLM)
- Join über `row_id`, eine CSV ohne Split

```python
import pandas as pd
```

## Cleaned-CSV & Rohtexte laden

```python
df = pd.read_csv("../../data/preprocessed/cleaned_fake_jobs.csv")

raw = pd.read_csv("../../data/raw/fake_job_postings.csv")
text_cols = ["title", "company_profile", "description", "requirements", "benefits"]
texts = raw[["job_id"] + text_cols].rename(columns={"job_id": "row_id"})
texts[text_cols] = texts[text_cols].fillna("")
```

## Texte per row_id anhängen & speichern

```python
out = df.merge(texts, on="row_id", how="left")
out.to_csv("../../data/preprocessed/cleaned_text_fake_jobs.csv", index=False)
```

## Verifikation

```python
print("shape", out.shape)
assert out[text_cols].isna().sum().sum() == 0
print(out[text_cols].head(2))
```

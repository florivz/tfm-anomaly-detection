# Fake Jobs – Cleaned + Text
- `row_id` + Label + **rohe Freitexte** aus den Rohdaten
- Seed-unabhängig: hier wird nichts gefittet, das Zeilen-Universum kommt aus `rows_fake_jobs.csv`

```python
import pandas as pd

TEXT_COLS = ["title", "company_profile", "description", "requirements", "benefits"]
```

## Zeilen-Universum & Rohtexte laden
- `rows_fake_jobs.csv` definiert die Zeilen (bei Airbnb nach ICC-Filter)

```python
rows = pd.read_csv("../../data/splits/rows_fake_jobs.csv")
raw = pd.read_csv("../../data/raw/fake_job_postings.csv")
raw["row_id"] = raw["job_id"].astype("int64")
texts = raw[["row_id"] + TEXT_COLS].copy()
texts[TEXT_COLS] = texts[TEXT_COLS].fillna("")
```

## Zusammenführen & speichern

```python
out = rows.merge(texts, on="row_id", how="left")
out[TEXT_COLS] = out[TEXT_COLS].fillna("")
out.to_csv("../../data/preprocessed/cleaned_text_fake_jobs.csv", index=False)
print("Shape:", out.shape)
```

## Verifikation

```python
assert len(out) == len(rows)
assert out["row_id"].is_unique
print(out[TEXT_COLS].head(2))
```

# Airbnb Paris – Semantisch (PCA 30)
- PCA auf 30 Komponenten über den Embedding-Block (gesamter Datensatz)
- Erklärte Varianz wird geprintet (Ziel > 80 %)

```python
import pandas as pd
from sklearn.decomposition import PCA
```

## Semantic-CSV laden & Embedding-Block isolieren

```python
df = pd.read_csv("../../data/preprocessed/semantic_airbnb_paris.csv")

emb_cols = [c for c in df.columns if "_emb_" in c]
other_cols = [c for c in df.columns if c not in emb_cols]
```

## PCA(30)

```python
pca = PCA(n_components=30, random_state=42)
red = pca.fit_transform(df[emb_cols])
print("explained variance (30 comps):", round(pca.explained_variance_ratio_.sum(), 4))
pca_cols = [f"pca_{i}" for i in range(30)]
```

## Zusammenbauen & speichern

```python
out = pd.concat([df[other_cols].reset_index(drop=True), pd.DataFrame(red, columns=pca_cols)], axis=1)
out.to_csv("../../data/preprocessed/semantic_pca30_airbnb_paris.csv", index=False)
print("shape", out.shape)
```

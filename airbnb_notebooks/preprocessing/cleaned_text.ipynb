# Airbnb Paris – Cleaned + Freitexte
- Cleaned-Features plus die 4 originalen Freitextspalten (für AnoLLM)
- Join über `row_id`, eine CSV ohne Split

```python
import pandas as pd
```

## Cleaned-CSV & Rohtexte laden

```python
df = pd.read_csv("../../data/preprocessed/cleaned_airbnb_paris.csv")

raw = pd.read_csv("../../data/raw/airbnb_paris.csv", low_memory=False)
text_cols = ["name", "description", "neighborhood_overview", "host_about"]
texts = raw[["id"] + text_cols].rename(columns={"id": "row_id"})
texts[text_cols] = texts[text_cols].fillna("")
```

## Texte per row_id anhängen & speichern

```python
out = df.merge(texts, on="row_id", how="left")
out.to_csv("../../data/preprocessed/cleaned_text_airbnb_paris.csv", index=False)
```

## Verifikation

```python
print("shape", out.shape)
assert out[text_cols].isna().sum().sum() == 0
print(out[text_cols].head(2))
```

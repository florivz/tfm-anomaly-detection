# Airbnb Paris – Cleaned + Text
- `row_id` + Label + **rohe Freitexte** aus den Rohdaten
- Seed-unabhängig: hier wird nichts gefittet, das Zeilen-Universum kommt aus `rows_airbnb_paris.csv`

```python
import pandas as pd

TEXT_COLS = ["name", "description", "neighborhood_overview", "host_about"]
```

## Zeilen-Universum & Rohtexte laden
- `rows_airbnb_paris.csv` definiert die Zeilen (bei Airbnb nach ICC-Filter)

```python
rows = pd.read_csv("../../data/splits/rows_airbnb_paris.csv")
raw = pd.read_csv("../../data/raw/airbnb_paris.csv", low_memory=False)
raw["row_id"] = raw["id"].astype("int64")
texts = raw[["row_id"] + TEXT_COLS].copy()
texts[TEXT_COLS] = texts[TEXT_COLS].fillna("")
```

## Zusammenführen & speichern

```python
out = rows.merge(texts, on="row_id", how="left")
out[TEXT_COLS] = out[TEXT_COLS].fillna("")
out.to_csv("../../data/preprocessed/cleaned_text_airbnb_paris.csv", index=False)
print("Shape:", out.shape)
```

## Verifikation

```python
assert len(out) == len(rows)
assert out["row_id"].is_unique
print(out[TEXT_COLS].head(2))
```
